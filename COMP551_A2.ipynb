# Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import zipfile
import urllib.request
import io

RANDOM_SEED=2026
rng = np.random.default_rng(RANDOM_SEED)
```

## Loading the Data

```python
url = "https://archive.ics.uci.edu/static/public/94/spambase.zip"
response = urllib.request.urlopen(url)
zip_file = zipfile.ZipFile(io.BytesIO(response.read()))
data = pd.read_csv(zip_file.open("spambase.data"),header = None).to_numpy()

X_all = data[:, :57]
y_all = data[:, 57]

print(X_all.shape, y_all.shape)
```

## Splitting the Data

```python
# 5% train, 95% test
#shuffle data
permutations = rng.permutation(len(y_all))
num_train = int(0.05 * len(y_all))
#print(num_train)

#split into rows of 230 and 4371
train_i = permutations[: num_train]
test_i = permutations[num_train: ]

X_train_raw = X_all[train_i]
y_train = y_all[train_i]

X_test_raw = X_all[test_i]
y_train = y_all[test_i]

print(X_train_raw.shape, X_test_raw.shape)
```

## Standardise

```python
# z-score = ( x - µ ) / std
def standardise(X_train_part, X_other_part):
  mean = X_train_part.mean(axis = 0)
  std = X_train_part.std(axis = 0)

  #avoid zero division
  std[std==0]=1

  return ((X_train_part - mean) / std), ((X_other_part - mean) / std)

X_train, X_test = standardise(X_train_raw, X_test_raw)
```
